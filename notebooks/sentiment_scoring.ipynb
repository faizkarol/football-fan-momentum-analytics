```python
import pandas as pd
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
```

```python
tweet_path = r"H:\FOM- Study Material\Semester Three\Big-Data-Analysis Project\fam_momentum_index\Data\Final_Aligned_Tweet_Dataset.csv"
df = pd.read_csv(tweet_path)
```

```python
# Initialize sentiment analyzer
analyzer = SentimentIntensityAnalyzer()

# Apply sentiment scoring
df["sentiment_score"] = df["text"].apply(lambda x: analyzer.polarity_scores(x)["compound"])

#Convert and round timestamps
df["timestamp"] = pd.to_datetime(df["timestamp"])
df["minute"] = df["timestamp"].dt.floor("min")
```

```python
#Group and aggregate
agg = df.groupby("minute").agg(
    tweet_volume=("text", "count"),
    avg_sentiment=("sentiment_score", "mean")
).reset_index()
```

```python
#Normalize
agg["norm_volume"] = (agg["tweet_volume"] - agg["tweet_volume"].min()) / (agg["tweet_volume"].max() - agg["tweet_volume"].min())
agg["norm_sentiment"] = (agg["avg_sentiment"] - agg["avg_sentiment"].min()) / (agg["avg_sentiment"].max() - agg["avg_sentiment"].min())
```

```python
# Save to CSV
output_path = r"H:\FOM- Study Material\Semester Three\Big-Data-Analysis Project\fam_momentum_index\Data\tweet_sentiment_by_minute.csv"
agg.to_csv(output_path, index=False)

print("Sentiment scoring and aggregation complete.")
```
